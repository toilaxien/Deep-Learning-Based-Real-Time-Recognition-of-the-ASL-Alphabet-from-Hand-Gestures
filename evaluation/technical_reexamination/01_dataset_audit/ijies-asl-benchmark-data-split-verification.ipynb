```python
import os
import shutil
import random
from tqdm import tqdm

# 📂 Thư mục gốc chứa ảnh
source_dir = "/kaggle/input/asl-alphabet/asl_alphabet_train/asl_alphabet_train"

# 📂 Thư mục đầu ra
output_root = "/kaggle/working/asl_split"
os.makedirs(output_root, exist_ok=True)

# 📊 Tỷ lệ chia
test_ratio = 0.2       # 20% test
val_within_train = 0.2 # 20% của phần còn lại (tức là 0.2 * 0.8 = 0.16 = 16% tổng)

# Tạo thư mục train/val/test
for split in ['train', 'val', 'test']:
    os.makedirs(os.path.join(output_root, split), exist_ok=True)

# Lặp qua từng lớp (A, B, C, ...)
for class_name in sorted(os.listdir(source_dir)):
    class_path = os.path.join(source_dir, class_name)
    if not os.path.isdir(class_path):
        continue

    # Lấy tất cả ảnh (định dạng phổ biến)
    images = [f for f in os.listdir(class_path)
              if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

    random.seed(42)
    random.shuffle(images)

    n_total = len(images)
    n_test = int(test_ratio * n_total)          # 20% test
    n_val = int(val_within_train * (n_total - n_test))  # 20% của 80% còn lại
    n_train = n_total - n_test - n_val

    train_imgs = images[:n_train]
    val_imgs = images[n_train:n_train + n_val]
    test_imgs = images[n_train + n_val:]

    # Tạo thư mục cho mỗi split
    for split, split_imgs in zip(['train', 'val', 'test'], [train_imgs, val_imgs, test_imgs]):
        split_class_dir = os.path.join(output_root, split, class_name)
        os.makedirs(split_class_dir, exist_ok=True)

        for img_name in tqdm(split_imgs, desc=f"{class_name} → {split}", leave=False):
            src = os.path.join(class_path, img_name)
            dst = os.path.join(split_class_dir, img_name)
            shutil.copy(src, dst)

print("✅ Đã chia dữ liệu thành công!")
print(f"📁 Kết quả lưu tại: {output_root}")
```

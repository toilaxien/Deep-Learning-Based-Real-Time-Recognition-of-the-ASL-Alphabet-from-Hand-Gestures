```python
from ultralytics import YOLO

model = YOLO("yolo12x.pt")
```

```python
#Train model
results = model.train(
    data="/kaggle/working/dataset.yaml",
    epochs=60,
    imgsz=224,
    batch=128,
    patience=5,
    lr0=0.001,
    weight_decay=0.0005,
    mosaic=1.0,
    mixup=0.1,
    hsv_h=0.015, hsv_s=0.7, hsv_v=0.4,
    degrees=10, translate=0.1, scale=0.5, shear=0.2,
    perspective=0.0005,
    flipud=0.0, fliplr=0.5,
    project="Train_results_YOLO12x_OD_ASL",
    exist_ok=True,
    device=[0,1]
)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11m-cls.pt")
```

```python
results = model.train(
    data="/kaggle/input/d/ngobinhxuyen/asl-alphabet/asl_split",  # Đường dẫn dataset (train/val/test hoặc các thư mục class)
    
    epochs=155,            # Số epoch huấn luyện tối đa
    imgsz=224,             # Kích thước ảnh đầu vào (224x224)
    batch=256,             # Batch size – số ảnh xử lý mỗi lần
    patience=10,           # Early stopping: dừng nếu val loss không cải thiện sau 10 epoch
    
    lr0=0.001,             # Learning rate ban đầu
    weight_decay=0.0005,   # Regularization L2, giảm overfitting
    
    mosaic=1.0,            # Bật mosaic augmentation (ghép 4 ảnh)
    mixup=0.1,             # MixUp: trộn 2 ảnh và nhãn (mức nhẹ)
    
    hsv_h=0.015,           # Thay đổi sắc độ màu (Hue)
    hsv_s=0.7,             # Thay đổi độ bão hòa màu (Saturation)
    hsv_v=0.4,             # Thay đổi độ sáng (Value)
    
    degrees=10,            # Xoay ảnh ngẫu nhiên ±10 độ
    translate=0.1,         # Dịch chuyển ảnh ±10%
    scale=0.5,             # Phóng to / thu nhỏ ảnh (50%)
    shear=0.2,             # Nghiêng ảnh (shear transformation)
    perspective=0.0005,    # Biến dạng phối cảnh (rất nhẹ)
    
    flipud=0.0,            # Không lật ảnh theo chiều dọc
    fliplr=0.5,            # Lật ảnh ngang với xác suất 50%
    
    project="/kaggle/working/runs",  # Thư mục lưu kết quả train
    name="gesture_train_v1",         # Tên experiment
    exist_ok=True,                  # Cho phép ghi đè nếu đã tồn tại
    
    device=[0,1]            # Sử dụng GPU 0 và 1 (multi-GPU)
)
```
